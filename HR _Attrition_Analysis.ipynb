# EDA

```python
df.head()
```

```python
sns.countplot(x= df.Department)
```

```python
Cat_col=[]
for i in df:
    if len(df[i].unique())<20:
        Cat_col.append(i)
```

```python
Cat_col
```

```python
df_cat= df[Cat_col]
df_cat.head()
```

```python
cont_col=[]
for i in df:
    if len(df[i].unique())>20:
        cont_col.append(i)
```

```python
cont_col
```

```python
df_cont= df[cont_col]
```

```python
df_cont.head()
```

# Univariate Analysis

```python
df_cat.head()
```

```python
plt.figure(figsize=(10,10))
plotnumber=1
for i in df_cat:
    if plotnumber<=4:
        plt.subplot(2,2,plotnumber)
        sns.scatterplot(x=df_cat[i],y=df_cat[i],hue=df['Attrition'])
        plotnumber=plotnumber+1
plt.tight_layout()
```

```python
plt.figure(figsize=(10,10))
plotnumber=1
for i in df_cont:
    if plotnumber<=4:
        plt.subplot(3,3,plotnumber)
        sns.histplot(x=df_cont[i])
        plotnumber=plotnumber+1
plt.tight_layout()
```

```python
sns.scatterplot(x=df['Age'],y=df['PercentSalaryHike'],hue=df['Attrition'])
```

```python
sns.pairplot(df[['PerformanceRating','WorkLifeBalance','YearsInCurrentRole','Department','Gender']],diag_kind='kde')
```

```python
plt.figure(figsize=(15,20))
plotnumber=1
target="Attrition"
for column in df_cont:
    if column !=target:
        plt.subplot(5,3,plotnumber)
        sns.boxplot(x=df[target],y=df[column],data=df)
        plt.title('{column} vs {target}')
        plotnumber+=1
plt.tight_layout()
plt.show()
```

# Bivariate Analysis

```python
plt.figure(figsize=(10,10))
plotnumber=1
for i in df_cat:
    if plotnumber<=4:
        plt.subplot(2,2,plotnumber)
        sns.countplot(x=df_cat[i],hue=df['Attrition'])
        plotnumber=plotnumber+1
plt.tight_layout()
```

```python
plt.figure(figsize=(10,10))
plotnumber=1
for i in df_cat:
    if plotnumber<=4:
        plt.subplot(2,2,plotnumber)
        sns.histplot(x=df_cat[i],hue=df['Attrition'])
        plotnumber=plotnumber+1
plt.tight_layout()
```
